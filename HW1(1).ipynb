## Solution

### 1. Sigmoid Function Gradient Derivation

We start with the sigmoid function:

$$
\sigma(x) = \frac{1}{1 + e^{-x}} = \left(1 + e^{-x}\right)^{-1}
$$

**Step 1:** Apply the chain rule. Let $u = 1 + e^{-x}$, so $\sigma(x) = u^{-1}$.

$$
\frac{d\sigma(x)}{dx} = \frac{d}{dx}\left(u^{-1}\right) \cdot \frac{du}{dx}
$$

**Step 2:** Compute $\frac{d}{dx}\left(u^{-1}\right)$:

$$
\frac{d}{dx}\left(u^{-1}\right) = -u^{-2} = -\frac{1}{u^2} = -\frac{1}{(1 + e^{-x})^2}
$$

**Step 3:** Compute $\frac{du}{dx} = \frac{d}{dx}(1 + e^{-x})$:

$$
\frac{du}{dx} = 0 + \frac{d}{dx}(e^{-x}) = e^{-x} \cdot \frac{d}{dx}(-x) = e^{-x} \cdot (-1) = -e^{-x}
$$

**Step 4:** Apply the chain rule:

$$
\frac{d\sigma(x)}{dx} = -\frac{1}{(1 + e^{-x})^2} \cdot (-e^{-x}) = \frac{e^{-x}}{(1 + e^{-x})^2}
$$

**Step 5:** Express in terms of $\sigma(x)$. Note that:

$$
\sigma(x) = \frac{1}{1 + e^{-x}} \quad \Rightarrow \quad 1 - \sigma(x) = 1 - \frac{1}{1 + e^{-x}} = \frac{1 + e^{-x} - 1}{1 + e^{-x}} = \frac{e^{-x}}{1 + e^{-x}}
$$

**Step 6:** Multiply $\sigma(x)$ and $(1 - \sigma(x))$:

$$
\sigma(x)(1 - \sigma(x)) = \frac{1}{1 + e^{-x}} \cdot \frac{e^{-x}}{1 + e^{-x}} = \frac{e^{-x}}{(1 + e^{-x})^2}
$$

**Step 7:** Compare with our result from Step 4:

$$
\frac{d\sigma(x)}{dx} = \frac{e^{-x}}{(1 + e^{-x})^2} = \sigma(x)(1 - \sigma(x))
$$

**✓ Therefore, we have proven:**

$$
\frac{d\sigma(x)}{dx} = \sigma(x)(1 - \sigma(x))
$$


### 2. Tanh Function Gradient Derivation

We start with the hyperbolic tangent function:

$$
\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}
$$

**Method 1: Direct Differentiation Using Quotient Rule**

**Step 1:** Apply the quotient rule: $\frac{d}{dx}\left(\frac{f(x)}{g(x)}\right) = \frac{f'(x)g(x) - f(x)g'(x)}{g(x)^2}$

Let $f(x) = e^x - e^{-x}$ and $g(x) = e^x + e^{-x}$.

**Step 2:** Compute $f'(x)$:

$$
f'(x) = \frac{d}{dx}(e^x - e^{-x}) = e^x - (-e^{-x}) = e^x + e^{-x} = g(x)
$$

**Step 3:** Compute $g'(x)$:

$$
g'(x) = \frac{d}{dx}(e^x + e^{-x}) = e^x + (-e^{-x}) = e^x - e^{-x} = f(x)
$$

**Step 4:** Apply the  rule:

$$
\frac{d\tanh(x)}{dx} = \frac{f'(x)g(x) - f(x)g'(x)}{g(x)^2} = \frac{g(x) \cdot g(x) - f(x) \cdot f(x)}{g(x)^2}
$$

$$
= \frac{g(x)^2 - f(x)^2}{g(x)^2} = \frac{(e^x + e^{-x})^2 - (e^x - e^{-x})^2}{(e^x + e^{-x})^2}
$$


**Step 5:** therefore:

$$
\frac{(e^x + e^{-x})^2 - (e^x - e^{-x})^2}{(e^x + e^{-x})^2} = 1 - \tanh(x)^2
$$
